```python
#Ejercicio 1
from typing import Any
def extract_contenido(ruta:str)-> dict[str,Any]:
    mi_lista=[]
    with open(ruta,"r") as f:
        contenido=f.read()
        lineas=contenido.split("\n")
        #Revisa linea por linea
        for linea in lineas:
            #Eliminar la primera linea que contiene lista
            if "registro" in linea.lower():
                continue
            #Eliminar lugar vacio
            if "id" in linea.lower():
                            continue
            if not linea :
                continue
            #Eliminar instrucciones mediante break, rompe el bucle que queda para el final
            if "instrucciones" in linea.lower():
                continue
            if linea.startswith("1."):
                break
            mi_lista.append(linea)
           
    return mi_lista

contenido=extract_contenido(
    "C:/Users/USER/Documents/DataScience/01-python-intro/data/ventas_polos_aqp.txt"
)
```

```python
class Producto:
    def __init__(
        self,
        ID:str,
        producto:str,
        talla:str,
        color:str,
        cliente:str,
        ciudad:str,
        cantidad:int,
        precio:float,
        metodo_pago:str,
        estado:str
                 )->None:
        self.ID=ID
        self.producto=producto
        self.talla=talla
        self.color=color
        self.cliente=cliente
        self.ciudad=ciudad
        self.cantidad=cantidad
        self.metodo_pago=metodo_pago
        self.estado=estado
        self.precio=precio

class Ventas:
    def __init__(self, ventas: list[str]) -> None:
        self.ventas: list[Producto] = []

        for vent in ventas:
            informacion = [item.strip() for item in vent.split(",")]

            producto = Producto(
                ID=informacion[0],
                producto=informacion[1],
                talla=informacion[2],
                color=informacion[3],
                cliente=informacion[4],
                ciudad=str(informacion[5]),
                cantidad=int(informacion[6]),
                precio=float(informacion[7]),
                metodo_pago=informacion[8],
                estado=informacion[9]
            )

            self.ventas.append(producto)

    def total_vendido(self):
        for venta in self.ventas:
            total = venta.cantidad * venta.precio
            print(f"{venta.ID}:  {total} soles ")

    def saber_estado(self,estado:str)->None:
        lista_estado=[]
        for venta in self.ventas:
            if venta.estado.lower()==estado:
                print(venta.cliente,venta.ID)
                lista_estado.append(venta)
        return lista_estado
                
    def estado_total(self,estado:str="entregado"):
        total=0
        for venta in self.ventas:
            if venta.estado.lower() == "entregado":
                total+=venta.cantidad*venta.precio
        print("El total vendido de entregados es:" ,{total})

    def cantidad_estado(self,estado:str):
            cantidad=len(self.saber_estado(estado))
            return print(f"La cantidad de productos {estado} son {cantidad}")
            
    def producto_mas_vendido(self):
        productos = {}
        for venta in self.ventas:
            if venta.producto in productos:
                productos[venta.producto] += venta.cantidad
            else:
                productos[venta.producto] = venta.cantidad

        producto_mayor = max(productos, key=productos.get)

        print("El producto más vendido es:", producto_mayor)
        

    def ciudad_ventas(self,ciudad:str):
        cantidad=0
        for venta in self.ventas:
            if venta.ciudad.lower()==ciudad:
                 cantidad+=1
        print("La cantidad vendida es",cantidad)
                
            

    def metodo_mas_usado(self):
        ventas_metodo={}
        for venta in self.ventas:
            if venta.metodo_pago in ventas_metodo:
                ventas_metodo[venta.metodo_pago] += venta.cantidad
            else:
                ventas_metodo[venta.metodo_pago] = venta.cantidad
        pago_mas_usado =max(ventas_metodo,key=ventas_metodo.get)
        print("El pago mas usado es",pago_mas_usado)
        

        

ventas = Ventas(contenido)
```
