```python
#Ejercicio 1
from typing import Any
def extract_contenido(ruta:str)-> dict[str,Any]:
    mi_lista=[]
    with open(ruta,"r") as f:
        contenido=f.read()
        lineas=contenido.split("\n")
        #Revisa linea por linea
        for linea in lineas:
            #Eliminar la primera linea que contiene lista
            if "lista" in linea.lower():
                continue
            #Eliminar lugar vacio
            if not linea :
                continue
            #Eliminar instrucciones mediante break, rompe el bucle que queda para el final
            if "instrucciones" in linea.lower():
                break
            mi_lista.append(linea)
           
    return mi_lista

contenido=extract_contenido(
    "C:/Users/USER/Documents/DataScience/01-python-intro/data/estudiantes_python.txt"
)
```

```python
class Estudiante:
    def __init__(
        self,
        nombre:str,
        edad:int,
        ciudad:str,
        nota:float
    )->None:
        self.nombre=nombre
        self.edad=edad
        self.ciudad=ciudad
        self.nota=nota

class Aula:
    def __init__(
        self,
        estudiantes:list[str])->None:
        self.estudiantes:list[Estudiante]=[]

        for estudiante in estudiantes:
            info=[item.strip()for item in estudiante.split(",")]
            estudiante=Estudiante(
                nombre=info[0],
                edad=int(info[1]),
                ciudad=info[2],
                nota=float(info[3])
            )
            self.estudiantes.append(estudiante)
            
    def __len__(self)-> int:
        return len(self.estudiantes)
    
    def filtro_por_ciudad(self,ciudad:str)-> None:
        for estudiante in self.estudiantes:
            if estudiante.ciudad.lower()==ciudad:
                print(estudiante.nombre)
    def get_aprobados(self,min_nota:int=13)->list[str]:
        estudiantes=[]
        for estudiante in self.estudiantes:
            if estudiante.nota >= min_nota:
                estudiantes.append(estudiante.nombre)
        return estudiantes
    def num_aprobados(self,min_nota:int=13)->int:
        return len(self.get_aprobados(min_nota))

    def export_alumnos(self,ruta:str)->None:
        with open(ruta,"w") as f:
            f.write("nombre,edad,ciudad,nota\n")

        for estudiante in self.estudiantes:
            with open(ruta,"a") as f:
                f.write(f"{estudiante.nombre},{estudiante.edad},{estudiante.ciudad},{estudiante.nota}\n" )

aula=Aula(contenido)
```
