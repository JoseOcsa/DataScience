## GRAFICO DE LINEAS

```python
from matplotlib  import pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[10,20,25,30,35]
plt.plot(x,y,marker='o',linestyle='-',color='b')
plt.xlabel('Periodo')
plt.ylabel('Valor')
plt.title('Ejemplo grafico de lineas')
plt.show()
```

## GRAFICO  DE BARRAS
SIRVE PARA COMPARAR MAGNITUDES Y RESPONDER PREGUNTAS COMO QUE CATEGORIA VENDE MAS ,QUE CLASE TIENE PROMEDIO MAS ALTO

```python
categorias=["A","B","C","D"]
valores=[40,70,30,90]
plt.bar(categorias,valores,color=['#D414E3','red','black','green'])
plt.xlabel('categorias')
plt.ylabel('frecuencia')
plt.title('grafico de barras verticales')
plt.show()
```

```python
plt.barh(categorias,valores,color='green')
plt.xlabel('frecuencia')
plt.ylabel('categoria')
plt.title('grafico de barras verticales')
plt.show()
```

## GRAFICO PASTEL O PIE CHART


```python
frutas=["manzanas","platano","peras"]
porcentajes=[30,50,20]
plt.pie(porcentajes,labels=frutas,autopct='%1.1F%%')
plt.title("grafico de pastel")
plt.show()
```

## GRAFICO DE DISPERSION (SCATTER PLOT) 
- este grafico me permite visualizar correlacion,fuerza y relacion  entre 2 variablees tambien puede mostrar valores atipicos

```python
import numpy as np
x=np.random.rand(50)
y=np.random.rand(50)
plt.scatter(x,y,color='blue')
plt.xlabel('variable x')
plt.ylabel('variable y')
plt.title('Dispersion')
plt.show()
```

## GRAFICO HISTOGRAMA 
- Reconocer concentracion ,tambien la dispersion y la asimetria. El numero de bins importa muy pocos quita detalle y muchos bins hacen ruido visual

```python
datos=np.random.rand(1000)
plt.hist(datos,bins=20,color='orange',edgecolor='black')
plt.xlabel('valor')
plt.ylabel('frecuencia')
plt.title('histograma') 
plt.show()
```

## GRAFICOS DE CAJAS,BIGOTES O BOXPLOT

```python
data_box=[np.random.normal(0,std,1000) for std in range(1,4)]
plt.boxplot(data_box,orientation='vertical', patch_artist=True)
plt.xlabel('grupo')
plt.ylabel('valor')
plt.show()
```
