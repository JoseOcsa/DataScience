Ejercicio 4. Resultados académicos


```python

notas = np.array([[14, 16, 13, 15], 
                  [18, 17, 16, 19], 
                  [11, 13, 12, 14], 
                  [16, 15, 17, 16], 
                  [13, 10, 15, 12]]) 
puntajes_total=notas.sum(axis=1)
print(puntajes_total)

promedio=notas.mean(axis=1)
print(promedio)

promedio_sin_mean= puntajes_total /  notas.shape[1]
print(promedio_sin_mean)
print("="*100)
mejor_estudiante= promedio.argmax()+1
print(mejor_estudiante)
print("="*100)
puntaje_total_materia=notas.sum(axis=0)
print(puntaje_total_materia)
```
