```python
import numpy as np
from sklearn.linear_model import LinearRegression
 
X = np.array([1, 2, 3, 4]).reshape(-1, 1)
y = np.array([2, 4, 5, 8])
 
modelo = LinearRegression()
modelo.fit(X, y)

print(modelo.coef_)        # [1.9]  la pendiente
print(modelo.intercept_)   # 0.0   el punto de partida
print(modelo.predict([[5]]))   # [9.5]
```
