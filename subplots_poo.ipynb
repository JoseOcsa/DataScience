```python
import numpy as np
from matplotlib import pyplot as plt
from matplotlib.figure import Figure
```

```python
flg=Figure()
```

```python
x=np.linspace(0,5,11)
y1=x**2
y2=x**3
```

```python
flg1=plt.figure()


dx=flg1.add_subplot()
dx.plot(x,y1,'ro-')
dx.set_title("funcion cuadratica")
plt.show()
```

```python
flg2=plt.figure()


dx=flg2.add_subplot()
dx.plot(x,y2,'bo-')
dx.set_title("funcion cuadratica")
plt.show()
```
