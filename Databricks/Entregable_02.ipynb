#1 Muestra nombre y precio de todos los productos,del mas caro al mas barato

```sql
SELECT nombre,precio 
FROM productos
ORDER BY precio DESC
```

#2 Lista de clientes de españa o mexico que empiece por M

```sql
SELECT *
FROM clientes
WHERE pais IN ('ES','MX') AND nombre like 'M%'
```

#3 Obtén los productos con precio entre 20 y 60 € que tengan stock disponible (> 0).

```sql
SELECT * 
FROM productos
```

```sql
SELECT nombre,precio,stock
FROM productos
WHERE (precio>20 AND PRECIO<=60) AND stock >0
```

# 4 Muestra los clientes sin teléfono, sustituyendo el NULL por «sin dato» (COALESCE).

```sql
SELECT nombre,telefono,
    COALESCE(telefono,'sin dato') AS datos_nuevos
FROM clientes;
```

#5 Lista los 10 productos más caros con su precio con IVA (× 1,21) usando un alias.

```sql
SELECT nombre,
        precio *1.21 AS precio_iva
FROM productos
ORDER BY precio_iva DESC
LIMIT 10
```

#6 Clasifica los productos en «barato», «medio» o «caro» con CASE.

```sql
SELECT nombre,precio,
    CASE 
        WHEN (precio<30) THEN 'Precio Barato'
        WHEN (precio>=30 AND precio <=60) THEN 'Precio Medio'
        ELSE 'Precio Caro'
    END AS Clasificacion_productos
FROM productos
ORDER BY precio ASC
```
