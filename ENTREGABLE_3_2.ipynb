9. Histograma: longitud del sépalo
 
Crea un histograma de `sepal_length_cm` con 10 bins, borde visible y etiquetas. Describe dónde se 
concentra la mayor parte de las observaciones.

```python
plt.hist(df["sepal_length_cm"],bins=10,color='yellow',edgecolor='black')
plt.xlabel('Longitud del sepalo')
plt.ylabel('Cantidad de flores')
plt.title('Histograma: Longitud del Sepalo')
plt.show()
```

Descripción: La mayor parte de las observaciones se concentra en el rango de 5.0 cm a 6.5 cm, teniendo un pico principal en el intervalo comprendido entre 5.5 y 6.0 cm.

10. Histograma y bins
 
Crea dos histogramas de `petal_length_cm`: uno con 5 bins y otro con 20 bins. Colócalos lado a lado 
usando subplots. Explica brevemente cómo cambia la interpretación al modificar los bins.

```python
plt.figure(figsize=(14, 5))
plt.subplot(1,2,1)
plt.hist(df["petal_length_cm"],bins=5,color='#52CCCC',edgecolor='black')
plt.xlabel('Longitud del sepalo')
plt.ylabel('frecuencia')
plt.title('Histograma: Longitud del Sepalo')


plt.subplot(1,2,2)
plt.hist(df["petal_length_cm"],bins=20,color='#52CCCC',edgecolor='black')
plt.xlabel('Longitud del sepalo')
plt.ylabel('Frecuencia')
plt.title('Histograma: Longitud del Sepalo')

plt.suptitle('Comparación de histogramas de longitud del sépalo mediante los bins', fontsize=16)

plt.tight_layout(w_pad=5)
plt.show()


```

La conclusion seria el aumento de detalle a la hora de distribucion de la longitud del sepalo,con 5 se muestra una tendencia general ,mientras que con 20 bins se muestra con mayor precision.

11. Boxplot por especie
 
Construye un boxplot de `sepal_length_cm` separado por especie. Etiqueta cada caja con el nombre de 
su especie y compara visualmente las medianas.

```python
## SABER las especies unicas de la lista
especies = df["species"].unique()
# usar una lissta vacia
datos = []

for especie in especies:
    grupo = df[df["species"] == especie]["sepal_length_cm"]
    datos.append(grupo)
#grfico usando para cambiar de color a las cajas, el relleno
# 
plt.boxplot(datos, tick_labels=especies,patch_artist=True,boxprops=dict(facecolor="lightblue"))
plt.xlabel("Especie")
plt.ylabel("Longitud del sépalo (cm)")
plt.title("Boxplot de sepal_length_cm por especie")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

12.  Boxplot de  variables numéricas
 
Crea un solo boxplot que compare las cuatro variables numéricas del dataset. Usa etiquetas cortas y 
legibles para cada caja. ¿Qué problema de escala observas?

```python
variables_numericas=df[["sepal_length_cm","sepal_width_cm","petal_length_cm","petal_width_cm"]]

etiquetas_numericas=["SepalLength","sepalWidth","PetalLength","PetalWidth"]


plt.boxplot(variables_numericas, tick_labels=etiquetas_numericas,patch_artist=True,boxprops=dict(facecolor="#56BA5F"))
plt.xlabel("Variables Numericas")
plt.ylabel("Medida en cm")
plt.title("Boxplot de variables numericas")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

Problema de escala: 

Se observa que las variables tienen rangos de valores diferentes. Las variables de longitud presentan valores más altos, mientras que petal_width_cm tiene valores mucho menores. Al mostrarlas en un mismo gráfico, la caja de petal_width_cm se ve más pequeña y resulta más difícil apreciar su distribución y variabilidad.

13. Subplots 1x2
 
Crea una figura con dos subplots: a la izquierda un scatter de `sepal_length_cm` vs. `sepal_width_cm`; a 
la derecha un histograma de `sepal_length_cm`. Agrega un título independiente a cada panel

```python
dashboard, axes = plt.subplots(1,2 , figsize=(16, 8))

#FIGURA 1
axes[0].scatter(df["sepal_length_cm"],df["sepal_width_cm"])
axes[0].set_xlabel('Longitud del sepalo')
axes[0].set_ylabel('Anchura del sepalo')
axes[0].set_title('Correlacion entre altura y anchura de un sepalo')
axes[0].grid(True, linestyle="--", alpha=0.5)

#FIGURA 2
axes[1].hist(df["sepal_length_cm"],bins=10,color='#52CCCC',edgecolor='black')
axes[1].set_xlabel('Longitud del sepalo')
axes[1].set_ylabel('Cantidad de flores')
axes[1].set_title('Histograma de sepalo')
axes[1].grid(True,linestyle="--", alpha=0.5)

dashboard.tight_layout(w_pad=4)
plt.show()
```
