```python
series.mean()
```

```python
series.median()
```

```python
series.mode()
```

```python
series.max()-series.min()
```

```python
q1=np.percentile(series,25)
print(q1)
q3=np.percentile(series,75)
print(q3)
```

```python
std=np.std(series)
std
```
